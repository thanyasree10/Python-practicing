## Python has two main types of conversion:
### 1. Implicit Type Conversion-Implicit means Python automatically converts a data type without asking the programmer to do it manually.
### 2. Explicit Type Conversion-the programmer manually converts one data type into another using conversion functions.

```python
product_1=100
product_2=0.55
total=product_1+product_2
print(total)
print(type(total))
```
