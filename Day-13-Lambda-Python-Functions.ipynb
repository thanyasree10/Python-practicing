# Lambda Python Functions
### lambda fucntion is small anonymous fucntion (a function without formal name)

```python
#syntax of lambda function
lambda arguments : expression
```

```python
#normal function 
def add(x,y):
    return x+y;
total=add(10,20)
print(total)
```

```python
add=lambda x,y: x+y #here a, b are parameter , a+b is expression
result=add(10,20) # arguments 10, 20
print(result)
#add variable name to stire the lambda fucntion
#lambda is keyword creating anonymous function
#a,b are parameter 
#a+b is expression

#Note: for lambda not required any return keyword becuase lambda automatically resturen expresisoin result 
```

```python
add=lambda x,y: x+y
add(10,20)
```

```python
square= lambda x : x*x
square(2)
```

```python
check=lambda x : "even" if x%2==0 else "odd"
print(check(5))
print(check(10))
```

```python
check = lambda x: "positive" if x>=0 else "negative"
print(check(8))
print(check(-7))
```

# eval
### built in function take string and evaluates expression

```python
num = eval(input("enter number"))
if num>10:
    print ("number is greater than 10")
else:
    print("numebr is less than 10")
```

# Walrus Opeartors
## walrus opeartor := is used to value to variable use same variabel i same expresison

```python
number=[10,20,30,40]
total=len(number)
if total>3:
    print(total)
```

```python
numbers=[10,20,30,40]
if (total := len(numbers)) > 3:
    print(total)
```
