## What is SDK
#### SDK is caollection of tool and libraries and functions provided by a company to help developres use to interact
#### SDK stands fro Software Development Kit
#### Suppose we want to communicate with Gemini AI from Python. We don't need to develop all the communication logic from scratch.Google provides a ready-made SDK called Google GenAI, We install that SDK using pip, import it into Python, create a client and call ready-made methods.That's the main purpose of an SDK: making developers' work easier by providing ready-made tools and functions.
#### Google provides a Python SDK called google-genai. -->pip install google-genai --> from google import genai
#### OpenAI provides an official Python SDK called openai. -->pip install openai --> from openai import OpenAI##
#### Groq provides a Python SDK called groq. --> pip install groq --> from groq import Groq

```python
from google import genai

client = genai.Client(api_key="YOUR_API_KEY")

interaction = client.interactions.create(
    model="gemini-3.5-flash",
    input="Explain python in two lines"
)

print(interaction.output_text)
```

#### Explanation:
######  1. from google import genai – Imports Google's Generative AI library.
###### 2. genai.Client(api_key=...) – Creates a client to communicate with Google's Gemini API using an API key.
###### 3. client.interactions.create() – Sends a request to the Gemini AI model.
###### 4. model – Specifies which Gemini model to use.
###### 5. input – Contains the prompt or question sent to the AI.
###### 6. interaction – Stores the response returned by the API.
###### 7. interaction.output_text – Extracts the generated text.
###### 8. print() – Displays the AI response on the screen.

```python
from google import genai
client = genai.Client(api_key="YOUR_API_KEY")
#first call
interaction1 = client.interactions.create(
    model="gemini-3.5-flash",
    input="what is aws in one line?"
)

print(interaction1.output_text)

#second call
interaction2 = client.interactions.create(
    model="gemini-3.5-flash",
    input="what is python in one line?"
)
print(interaction2.output_text)


#third call
interaction3 = client.interactions.create(
    model="gemini-3.5-flash",
    input="what is java in one line?"
)
print(interaction3.output_text)
```

## with function

```python
from google import genai
client=genai.Client(api_key="YOUR_API_KEY")
def ask_gemini(prompt):
    interaction=client.interactions.create(
        model="gemini-3.5-flash",
        input=prompt
    )
    return interaction.output_text
result=ask_gemini("what is python in one line")
print(result)
```

```python
result1 = ask_gemini("what is python in one line?")
result2 = ask_gemini("what is java in one line?")
result3 = ask_gemini("what is aws in one line?")
print(result1)
print(result2)
print(result3)
```
