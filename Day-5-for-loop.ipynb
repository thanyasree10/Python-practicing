```python
name=['aws','azur','gcp']
for i in name:
    print(i)
```

```python
a=[1,2,3,4,5,6]
for i in a:
    if i==3:
        break;
    print(i)
```

```python
number={
    'name' :'thanya',
    'age' :'21'
}
for i in number.items():
    print(i)
    
```

```python
number={
    'name' :'thanya',
    'age' :'21'
}
for i in number.values():
    print(i)
```

```python
number={
    'name' :'thanya',
    'age' :'21'
}
for i in number.keys():
    print(i)
```

```python
total=1
dev= range(1,6)
for i in dev:
    total=total+1;
    
print("total=",total)
```

```python
x=[1,2,3,4,5,6]
for i in x:
    if i%2==0:
        print(i ," is even")
    else:
        print(i, " is odd")
        
```

```python
a=[1,5,2,7,9,6]
for i in a:
    if i==7:
      print(i," found")
      break;
```

```python
import random
for i in range(1,6):
    num=random.randint(1,100)
    if num%2==0:
        print(num,"even number" " iteration",i)
    else:
        print(num,"odd number" " iteration" ,i)
```
