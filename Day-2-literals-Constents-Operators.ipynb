## Operators
### 1) Arithmetic Operators (+, -, /, //, %, **)
### 2) Assignment Operators(=, +=, -=, *=, /=)
### 3) Comparision operators((==, !=, >, <, >=, <=)
### 4) Logical Operators(and ,or ,not)
### 5) Bitwise operator(&, |, ^, ~, <<, >>)
### 6) Identity Operators(is, is not)
### 7) Membership Operator(in ,not in)


```python
### Arithemetic operator
a = 20
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b) # quotient
print("Modulus:", a % b)  #remainder
print("Floor Division:", a // b) #Floor divison
print("Exponentiation:", a ** b) #square
```

# Camparison
### Comparison operators are used to compare two values or variables. They return a Boolean result: True or False.
#### Operator	  Name	                        Meaning
####   ==	     Equal to	                 Checks whether two values are equal
####   !=	     Not equal to	             Checks whether two values are different
####   >	     Greater than	             Checks whether left value is greater
####   <	     Less than	                 Checks whether left value is smaller
####  >=	     Greater than or equal to	 Checks whether left value is greater or equal
####  <=	     Less than or equal to	     Checks whether left value is smaller or equal

# Logical Operator
####  Operator	 Meaning	       When does it return True?
####   and	    Both conditions	    Both are True
####   or	    Either condition	At least one is True
####   not	    Reverse condition	Original condition is False

# Bitwise Operator
## Operator	Name	Purpose
## &	Bitwise AND	Returns 1 if both bits are 1
## \|	Bitwise OR	Returns 1 if at least one bit is 1
## ^	Bitwise XOR	Returns 1 if bits are different
## ~	Bitwise NOT	Inverts the bits
## <<	Left shift	Shifts bits to the left
## >>	Right shift	Shifts bits to the right

```python
##Left shift operator
a = 5

print("Original number:", a)
print("Binary number:", bin(a))

print("Left shift by 1:", a << 1)
print("Left shift by 2:", a << 2)
print("Left shift by 3:", a << 3)
```

```python
#Right shift
a = 8
print(a >> 1)
```

```python
#is operator
#The is operator checks whether two variables refer to the same object.
a = 10
b = 10

print(a is b)
```

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(a == b)
print(a is b)
```

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(a is not b)
#They have equal values, but they are different objects.
```

## in operator
### The in operator checks whether an element is present in a collection.

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
print("orange" in fruits)
```
