# List
### Characteristics of a List
### - Ordered – Maintains insertion order.
###  - Indexed – Supports positive and negative indexing.
### - Slicing – Supports slicing.
### - Mutable – We can add, update and remove elements.
### - Duplicates – Allows duplicate values.
### - Different data types – Can store different types together

```python
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print(fruits[0])
print(fruits[2])
print(fruits[-1])
print(fruits[1:4])
print(fruits[::-1])
```

```python
fruits = ["apple", "banana", "orange"]
fruits[0] = "mango"
print(fruits)
```

```python
numbers = [10, 20, 30, 20, 40]

print(len(numbers))       # Number of elements
print(numbers.count(20))  # Count occurrences
print(numbers.index(30))  # Find index
```

```python
fruits=["mango","apple","banana"]
print("original list:",fruits)
fruits.append("orange")
print(fruits)
fruits.extend(["grapes","kiwi"])
print(fruits)
fruits.insert(1 ,"cherry")
print(fruits)
fruits.remove("banana")
print(fruits)
```

## Set: A set stores unique elements.
### Does not allow duplicate values.
Does not guarantee insertion order.
Does not support indexing.
Does not support slicing.
Mutable.
Supports mathematical set operations.

```python
numbers = {10, 20, 30, 40,20}

numbers.remove(20)

print(numbers)
```

```python
numbers = {10, 20, 30,100}

numbers.discard(100)

print(numbers)
```

```python
numbers = {10, 30, 40,}

numbers.remove(20)

print(numbers)
```

```python

# Creating a dictionary
student = {
    "name": "Veera",
    "course": "FDE",
    "age": 30
}

print("Original Dictionary:", student)


# 1. keys() - Returns all keys
print("\n1. keys():")
print(student.keys())


# 2. values() - Returns all values
print("\n2. values():")
print(student.values())


# 3. items() - Returns key-value pairs
print("\n3. items():")
print(student.items())


# 4. get() - Accesses a value using a key
print("\n4. get():")
print(student.get("name"))
print(student.get("city"))


# 5. update() - Updates existing values or adds new pairs
print("\n5. update():")
student.update({"age": 25, "city": "Hyderabad"})
print(student)


# 6. pop() - Removes a specified key and returns its value
print("\n6. pop():")
removed_value = student.pop("age")
print("Removed value:", removed_value)
print(student)


# 7. popitem() - Removes the last inserted key-value pair
print("\n7. popitem():")
removed_item = student.popitem()
print("Removed item:", removed_item)
print(student)


# 8. setdefault() - Returns value if key exists,
# otherwise inserts the key with a default value
print("\n8. setdefault():")
student.setdefault("country", "India")
student.setdefault("name", "Unknown")
print(student)


# 9. copy() - Creates a shallow copy
print("\n9. copy():")
new_student = student.copy()
print("Copied dictionary:", new_student)


# 10. clear() - Removes all key-value pairs
print("\n10. clear():")
new_student.clear()
print("After clear:", new_student)


# 11. fromkeys() - Creates a new dictionary
# using given keys and a common value
print("\n11. fromkeys():")
keys = ["name", "age", "city"]
new_dict = dict.fromkeys(keys, "Unknown")
print(new_dict)
```
