## Python Exception Handling
#### try, except, else, finally , raise
#### alone try or exception block not allowed

```python
a=10
b=0
print(a/b)
```

```python
try:
    a=10;
    b=0;
    print(a/b)
except:
    print("cannot be divided with zero")
```

```python
try:
    a=10;
    b=0;
    print(a/b)
except Exception as e:
    print(e)
```

```python
try:
    a = 10
    b = 0
    print(a / b)

except ZeroDivisionError as e:
    print("Cannot divide by zero:", e)
```

```python
count = 1

while count <= 3:
    try:
        a = 10
        b = 1
        print(a / b)
        count += 1

    except Exception as e:
        print(e)
```

```python
count = 1

while count <= 3:
    try:
        a = 10
        b = 0
        print(a / b)
        count += 1

    except Exception as e:
        print(e)
        count += 1
```

## finally block


```python
try:
    a = 10
    b = 0
    print(a / b)

except:
    print("cannot divide by zero")

finally:
    print("execution completed")
```

```python
try:
    a = 10
    b = 2
    print(a / b)

except:
    print("Something went wrong")

finally:
    print("execution completed")
```

```python
try:
    file = open("data.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File not found")

finally:
    print("Program execution completed")
```

## else block
#### You haven't included an else example in your notebook yet. Let's understand it.
#### The else block executes only when the try block completes successfully without any exception.

```python
try:
    a = 10
    b = 2
    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Division successful")
```

```python
age=12
try:
    if age<18:
        raise Exception("not eligible for your age")
    print("eligible")
```

```python
age = 21

if age < 18:
    raise Exception("You must be 18 or older")

print("You are eligible")
```

```python
try:
    a= int(input("enter number for a"))
    b= int(input("enter number for b"))
    result = a/b
    print(result)
except ZeroDivisionError:
    print("cannot divide by zero")
except ValueError:
    print("please enter correct number only")
except TypeError:
    print("invalide data type")
except Exception as e:
    print("unexpecred error:", e)
```

```python
try:
    a= int(input("enter number for a"))
    b= int(input("enter number for b"))
    print(a/b)

except (ValueError,ZeroDivisionError) as e:
    print("please enter correct number only", e)
else:
    print("else excecution")
finally:
    print("program exceution completed")
```
