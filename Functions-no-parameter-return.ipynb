### Types of functions in Python
### Built-in functions - Built-in functions are predefined functions already available in Python.
### User-defined Functions-User-defined functions are functions created by programmers using the def keyword to perform a specific task.

```python
import builtins
print(dir(builtins))
```

## Four types of user-defined functions
### No parameters – No return value  --->Takes no input and returns no result. 
### No parameters – With return value --->Takes no input but returns a result.
### With parameters – No return value ---->Takes input but does not return a result.
### With parameters – With return value ---->Takes input and returns a result.
###
