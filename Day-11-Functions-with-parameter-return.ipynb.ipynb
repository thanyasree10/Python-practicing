# Function with parameters and return value
### First, understand two important terms.
### - Parameter: A variable mentioned inside the function definition.
### - Argument: The actual value passed when calling the function.
### - Return: Sends the result back to the place where the function was called.
### Example 1: Basic function

```python
def add(a,b):
    return a+b;
sum=add(10,20)
print(sum)
```

```python
def add(a,b):
    sum=a+b;
    return sum;
sum=add(10,20)
print(sum)
```

```python
def add(a,b):
    return a+b;
def multi(number):
    return number*2;
total=add(10,20)
final=multi(total)
print(final)
```

```python
def calculate_bill(price,quantity,discount):
    total=price*quantity
    discount_amount=total*discount/100
    total_amount=total-discount_amount
    return total_amount,discount_amount,total;
final=calculate_bill(1000,3,10)
print(final)
```

### There are six concepts:
##### 1. Positional arguments
##### 2. Keyword arguments
##### 3. Default arguments
##### 4. Variable-length positional arguments (*args)
##### 5. Keyword-only arguments
##### 6. Variable-length keyword arguments (**kwargs)

```python
#position arguments
def student(name, age, course):
    print(name)
    print(age)
    print(course)

student("Veera", 20, "FDE")
```

```python
#keyword arguments
def students(name,age,course):
    print(name)
    print(age)
    print(course)
student(age=20,name="Thanya",course="Multi-Cloud")
```

```python
#Default arguments
def student(name, age=21, city="Hyderabad"):
    print(name)
    print(age)
    print(city)

student("Thanya")
student("Akash", city="Chennai")
```

```python
#Variable-length positional arguments (*args)
def students(*names):
    return names;
students("Thanya","Meghana","sai sree","geetha")
```

```python
def students(*names):
    for i in names:
        return names;
student("thanya","Meghana","sai sree")
```

```python
#Keyword-only arguments
def employee(name, *, salary, city):
    print(name)
    print(salary)
    print(city)

employee("Veera", 50000, city="Hyderabad")
```

```python
def employee(name, *, salary, city):
    print(name)
    print(salary)
    print(city)

employee("Veera", salary=50000, city="Hyderabad")
```

```python
#kwargs – Variable-length keyword arguments
def student(**details):
    print(details)
student(
    name="Thanya",
    age=22,
    course="fde")
```

```python
def student(**details):
    print(details["name"])
    print(details["age"])
    print(details["city"])

student(name="Veera", age=30, city="Hyderabad")
```

```python
#using loop
def student(**details):
    for key,value in details.items():
        print(key, ":", value)
student(
    name="Thanya",
    age=21,
    course="fde"
)
```

```python
def student(*marks ,**details):
    return marks,details;
student(
    30 ,59, 44,
    name="Thanya",
    age=23
)
```

```python
#name ---> Accepts one positional value ---> String 
#*marks   -->Collects multiple positional values -->Tuple 
#**details --> Collects multiple keyword values ---> Dictionary 
```

## Function with Local and global variable
### global variable : define outside of the function
### local variable : define inside of the function

```python
name="FdE"
def course():
    print("course_name",name)
course()
```

```python
name="FDE"
def course():
    return name;
course()
```

```python
name = "FDE"  # Global variable

def naresh():
    name = "Veera"  # Local variable
    print("Hello", name)

naresh()
print(name)
```

```python
#global keyword
salary = 5000;
def salary_1():
    global salary
    salary=6000;
```
